# Nigerian Music scraped from Spotify - an analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
music = pd.read_csv('../data/nigerian-songs.csv')
music.head()
```

```python
music.info()
```

```python
music.isnull().sum()
```

```python
music.describe()
```

```python
new_music = music[music['popularity'] > 0]
new_music.describe()
```

```python
top = new_music['artist_top_genre'].value_counts()
plt.figure(figsize=(10,7))
sns.barplot(x=top[:5].index, y=top[:5].values)
```

```python
clean_music = new_music[new_music['artist_top_genre'] != 'Missing']
clean_music.describe()
```

```python
corrmat = clean_music.corr(numeric_only=True)
sns.heatmap(corrmat, vmax=.8, square=True)
```

```python
sns.set_theme(style="ticks")
g = sns.jointplot(
    data=clean_music,
    x="popularity", y="danceability", hue="artist_top_genre",
    kind="kde"
)
```

```python
sns.FacetGrid(clean_music, hue='artist_top_genre', height=5) \
   .map(plt.scatter, "popularity", "danceability") \
   .add_legend()
```

## ASSIGNMENT LESSON 14

## Experiment 1 (Energy & Loudness)

Here we are exploring the relationship between Energy and Loudness because on the heat map they showed a strong positive correlation.

```python
sns.FacetGrid(clean_music, hue='artist_top_genre', height=5) \
   .map(plt.scatter, "energy", "loudness") \
   .add_legend()
```

## Observation- Experiment 

The pattern looks clean and displays a strong positive relation between energy and loudness across the graph for all the geners. The geners a heavily overlapped aka mixed.

It would be difficult for clustering to differintiate between the two, when using unsupervised clustering.

## EXPERIMENT 2 (LENGTH & Release_Date)

Here we are exploring the relationship between Length and Release_Date because they show a strong negative realtion between them on the heat map.

```python
sns.FacetGrid(clean_music, hue='artist_top_genre', height=5) \
   .map(plt.scatter, "length", "release_date") \
   .add_legend()
```

## Observation- Experiment 2
