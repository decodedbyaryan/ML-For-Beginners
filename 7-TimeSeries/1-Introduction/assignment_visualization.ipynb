## Import
Importing the required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

### Dataset 1: *Daily Climate*

Loading the dataset

df_climate = pd.read_csv('data_assignment/DailyClimate_Train.csv')
df_climate['date'] = pd.to_datetime(df_climate['date'])
df_climate.set_index('date', inplace=True)
print(df_climate.head())

### Plotting
Now plotting the dataset to make the observation

```python
df_climate['meantemp'].plot(figsize=(15, 6), fontsize=12)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Mean Temperature (°C)', fontsize=12)
plt.title('Daily Mean Temperature 2013-2017')
plt.show()
```

### **Observation**
we can see few characteristics of the dataset that standout and match the features of time series, namely-

**Seasonality-** 

***Yearly cycle-*** We see that there is an yearly cycle where the temperature is low at the beginning of the year and it starts rising till the mid of the year after which is slowing start falling down again till the end of the year i.e. also the beginning of the new year.

***mid year trend-*** We can observe that during every mid of year there is a trend of temperature going to its peak (june-july, hot months).

***year end/beginning trend-*** We also observe that there is a trend of the tempearature falling to its lowest on our chart at the end of the year or the beginning of the new year.

**Constant Variance-**

The temperature keeps fluctuating daily, all around the year but it doesn't cross above or below a certain extent on the daily basis as it fluctuate.

**No long term trend-**

We can clearly see that there is no long term trend as the cycle kept repeating at the same levels across all four cycle on the plot.

### Dataset 2: *UAE Beverages*

Loading the dataset and summing up the net sales on the daily basis

```python
df_uae = pd.read_csv('data_assignment/UAE_Beverages_Sales_5000.csv')
df_uae['Date'] = pd.to_datetime(df_uae['Date'])
df_uae_daily = df_uae.groupby('Date')['Net_Sales'].sum().reset_index()
df_uae_daily.set_index('Date', inplace=True)
print(df_uae_daily.head())
```

### Plotting
Now plotting the dataset to make the observation

```python
df_uae_daily['Net_Sales'].plot(figsize=(15, 6), fontsize=12)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Net Sales (AED)', fontsize=12)
plt.title('UAE Beverages Daily Net Sales 2020-2025')
plt.show()
```

### **Observation**
we can see few characteristics of the dataset that standout and match the features of time series, namely-

**outliers-** 

There are outliers in the dataset i.e. the sharp upward spikes in the plot that are totally outside the trend and even the ones that are minor but standout as they are reaching close to zero on the plot.

**Possible long term trend-**

We can see how densly packed the data is and in this densly packed data there is a little difference where the over all data is higher in the early year i.e. 2020-2021 while it stays lower or never reaches the same height constantly and densely.

**Noisy constant variance-**

The fluctuating keeps happening throughout the data making is messay and not so easy to read.

### Dataset 3: *CPI Inflation Data*

Loading the data

```python
df_cpi = pd.read_csv('data_assignment/cpi_inflation.csv')
df_cpi['date'] = pd.to_datetime(df_cpi['date'])
df_cpi.set_index('date', inplace=True)
print(df_cpi.head())
```

### Plotting
Now plotting the dataset to make the observation

```python
df_cpi['us_cpi_all_items'].plot(figsize=(15, 6), fontsize=12)
plt.xlabel('Date', fontsize=12)
plt.ylabel('CPI Value', fontsize=12)
plt.title('US CPI Inflation 2015-Present')
plt.show()
```

### **Observation**
we can see few characteristics of the dataset that standout and match the features of time series, namely-

**Trend-** 

There is an overall trend here in the whole plot where the inflation keeps rising in the long term never showing the reverse.

**Abrupt Change-**

we see a sudden dip in 2020 where the trend is broken for the first time and it dips a little.

**Trend acceleration-**

From 2020 till mid of 2022, we see a sharp rise i.e. acceleration in the trend before it stablizing, moving into its normal speed after that.
