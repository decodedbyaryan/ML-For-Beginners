## Selected dataset
Today, We will be working on movie reviews dataset because these a text heavy, have clear positive and negative label and most importantly, the reviews are not straight out clear, they contain sarcasm, complex language which would give us some decent output to see how VADER works here along with a fun and clear comparison.

### Import
Importing all required libraries

```python
import pandas as pd
import nltk
import time
from nltk.corpus import stopwords
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from sklearn.datasets import fetch_20newsgroups
nltk.download('stopwords')
nltk.download('vader_lexicon')
```

### Data
Loading dataset

```python
movies = fetch_20newsgroups(subset='all', categories=['rec.sport.hockey', 'sci.space'],
                           remove=('headers', 'footers', 'quotes'))
df = pd.DataFrame({'text': movies.data, 'label': movies.target})
print(df.shape)
print(df['label'].value_counts())
```

### Inspect 
We are going to inspect what the data actually looks like plan our next set of actions

```python
print(df['text'][0][:500])
print("---")
print("Label:", df['label'][0])
```

### Removing stop words
Removing stop words to improve the processing time

```python
start = time.time()
cache = set(stopwords.words("english"))

def remove_stopwords(text):
    if not isinstance(text, str):
        return ""
    text = " ".join([word for word in text.split() if word not in cache])
    return text

df['text'] = df['text'].apply(remove_stopwords)
end = time.time()
print("Stop words removed in", round(end - start, 2), "seconds")
```

### Running VADER 
Now running VADER sentiment analysis on the posts

```python
vader_sentiment = SentimentIntensityAnalyzer()

def calc_sentiment(text):
    if not isinstance(text, str) or text == "":
        return 0
    return vader_sentiment.polarity_scores(text)["compound"]

start = time.time()
df["sentiment_score"] = df["text"].apply(calc_sentiment)
end = time.time()
print("Sentiment calculated in", round(end - start, 2), "seconds")
print(df[["text", "sentiment_score"]].head(5))
```

### Observation
The output visible here shows us that VADER did what we had previously concluded i.e. it marked the sentences solely based on the individual words in it and the total weightage of those words total to determine whether it is positive or negative or neutral sentence., as such without the context it totally misjudged the sentences that were sarcastic or meant something totally different. One additional, we also saw how it wasn't able to determine or couldn't analyze the sentences if they were too short

### Saving the result
