### NLTK Concept- Concordance
We would be observing how NLTK's concordance is used

```python
text1.concordance("Whale")
```

**Concordance --** Concordance helps find target word or words in the text and it also shows what text came before and after the target word while telling also mention the total times the target word or words appered.

#### Observation
The interesting part about the result was that the target word i.e. "whale" here was shown at the same position in the result, which made it a bit more easier to locate it and see what was around it at the same time.

### NLTK Concept- Similar words
We would be observing how NLTK's similar words concept is used

```python
text1.similar("whale")
```

**Similar words --** Similar words concept here in NLTK helps find words that are positioned often in the same grammatical way or position as the targeted word

#### Observation
The words that appeared in the result are different but maybe were used in similar kind of manner with for example they had "the" before them followed by a *verb*, etc.

*Mind you these are not similar to synonymous in any way*

### NLTK Concept- Frequency Distributions
We would be observing how NLTK's Frequency Distributions concept is used

```python
fdist1 = FreqDist(text1)
print(fdist1.most_common(20))
```

**Frequency Distribution --** Finds out how often and how many words appers in the text

#### Observation
We could see all the words at a glance in the result along with their frequecy, as such we can easily filter out the stopwords which are common, not very useful words from our text so that the NLP can understand the context better without having to go through every single word in the text, saving time and computing power.

*Without understanding why we needed this, it is very easy or common to determine frequency distribution concept as useless- I also did that before know why we need or use this*

### Visualizing Frequency Distribution
We would be observing how NLTK's Frequency Distributions concept is used in a visual format to get a better idea on how things work.

```python
fdist1.plot(20, cumulative=False)
```

**Visualizing Frequency Distribution --** .plot here helped us visualize the frequency distribution for the first 20 words.

**.plot --** was used here to display the graph above

#### Observation
The first two words or rather the first two samples on the graph i.e. "," and "the" appeared much more than any other sample starting from "."

The graph shape here tells us how some words appear to frequently while some might only do so once or twice, showing unbalanced things can get if we just go by the frequency without filtering the noise, *in this case the stopwords*.
