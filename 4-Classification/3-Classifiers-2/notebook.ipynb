# Build Classification Model

```python
import pandas as pd
cuisines_df = pd.read_csv("../data/cleaned_cuisines.csv")
cuisines_df.head()
```

```python
cuisines_label_df = cuisines_df['cuisine']
cuisines_label_df.head()
```

```python
cuisines_features_df = cuisines_df.drop(['Unnamed: 0', 'cuisine'], axis=1)
cuisines_features_df.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.linear_model import LogisticRegression 
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report
import numpy as np
```

```python
X_train, X_test, y_train, y_test = train_test_split(cuisines_features_df, cuisines_label_df, test_size=0.2)
```

```python
C = 10
classifiers = {
    'Linear SVC': SVC(kernel='linear', C=C, random_state=0),
    'KNN classifier': KNeighborsClassifier(C),
    'SVC': SVC(),
    'RFST': RandomForestClassifier(n_estimators=100),
    'ADA': AdaBoostClassifier(n_estimators=100)
}
```

```python
n_classifiers = len(classifiers)

for index, (name, classifier) in enumerate(classifiers.items()):
    classifier.fit(X_train, np.ravel(y_train))

    y_pred = classifier.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print("Accuracy (train) for %s: %0.1f%%" % (name, accuracy * 100))
    print(classification_report(y_test, y_pred))
```

# Parameter Play Assignment — AdaBoost Classifier

Which classifier you chose- AdaBoost

What your starting hypothesis is- reducing the n_estimators might help increase the accuracy because this would reduce the number of times ada runs (marking down its obsession with mistakes) and there is a chance that it would have a positive impact on the result unless we accidently reduce the n_estimators so much that it doesn't give ada the time and numbers needed to read the patterns, meaning lack in train = disasterous results


## Experiment 1: Baseline (n_estimators=100)

```python
ada_baseline = AdaBoostClassifier(n_estimators=100)
ada_baseline.fit(X_train, np.ravel(y_train))
y_pred = ada_baseline.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Baseline AdaBoost Accuracy: %0.1f%%" % (accuracy * 100))
print(classification_report(y_test, y_pred))
```

## Experiment 2: Reduced n_estimators (n_estimators=50)
Hypothesis: Fewer estimators means less obsession over noisy misclassified points, potentially improving accuracy.

```python
ada_exp2 = AdaBoostClassifier(n_estimators=50)
ada_exp2.fit(X_train, np.ravel(y_train))
y_pred = ada_exp2.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("AdaBoost n_estimators=50 Accuracy: %0.1f%%" % (accuracy * 100))
print(classification_report(y_test, y_pred))
```

## Result

Unfortunately, the results were disappointing and the AdaBoost accuracy with n_estimators=50 droped to 65.3% when compared to the original which was 67.7%, where n_estimators=100.

It seems like it was the result of adding not getting adiquate number of simulations or test runs during training that has resulted to this result.

AdaBoost builds each estimator sequentially to correct previous mistakes- with only 50 rounds, it didn't have enough correction cycles to learn the complex patterns in the cuisine data.

## Experiment 3: Increased n_estimators (n_estimators=200)
Hypothesis: More estimators means more correction cycles- but risks over-obsessing on noisy points.

```python
ada_exp3 = AdaBoostClassifier(n_estimators=200)
ada_exp3.fit(X_train, np.ravel(y_train))
y_pred = ada_exp3.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("AdaBoost n_estimators=200 Accuracy: %0.1f%%" % (accuracy * 100))
print(classification_report(y_test, y_pred))
```

## Result of Experiment 3
Suprisingly the accuracy increased to 69.5% with n_estimators=200, and it is higher than both the original 67.7% (n_estimators=100) and Experiment 2: 65.3% (n_estimators=50), thus proving my theory wrong and infact bearing the opposite results, where I assumed that reducing the n_esimators would result in better results because I was assuming that the lower results was the outcome of high number of n_estimators

## Final Verdict
Obsession in AdaBoost can only be seen when the n_estimators are very high (more than 500-1000), otherwise each run only increase the accuracy thus proving that more runs in AdaBoost bear good results until or unless they are too many which is much higher in number (generally above 500-1000).

However, this conclusion is specific to this cuisine dataset- noisier or cleaner dataset may show different sweet spots for n_estimators.

## Experiment 4
Let's try lowering ada's learning rate to 0.5 combined with our best n_estimators=200

```python
ada_exp4 = AdaBoostClassifier(n_estimators=200, learning_rate=0.5)
ada_exp4.fit(X_train, np.ravel(y_train))
y_pred = ada_exp4.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("AdaBoost lr=0.5, n=200 Accuracy: %0.1f%%" % (accuracy * 100))
print(classification_report(y_test, y_pred))
```

## Result Experiment 4
Reducing the learning rate has resulted in a negative impact on the accuracy which before was 69.5% in Experiment3 where the learning rate was normal/standard (i.e. 1) while it got low 68.2% here is Experiment 4 where learning rate was reduced to half of it normal rate to 0.5 here.

## Conclusion
While reducing the learning rate increases stability for AdaBoost, it at the same time impacts the speed of adaboost as such increases the need for more runs to get a better result than the one running at normal rate. Meaning you have to sacrifice speed for more stability but at the same time would need to run more cycles (n_estimators), to get better result
